```python
# NP42. Detect and replace outliers using NumPy.
df = pd.DataFrame({'Values': [10, 12, 14, 100, 15, 13, 11, 200, 16, 18]})
arr = df['Values'].values
Q1 = np.percentile(arr, 25)
Q3 = np.percentile(arr, 75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df['Clipped_Values'] = np.clip(arr, lower, upper)
print("Outlier detection and clipping:\n", df)
```

```python
# NP49. Perform a mini data analysis project.
np.random.seed(42)
n = 1000

# Step 1: Generate data
ages = np.random.randint(18, 65, n)
salaries = np.random.randn(n) * 20000 + 50000
scores = np.random.randn(n) * 10 + 75

df = pd.DataFrame({
    'Age': ages,
    'Salary': salaries,
    'Score': scores
})

# Step 2: Clean data
df = df[df['Salary'] > 0]
df['Salary'] = np.clip(df['Salary'], 20000, 150000)

# Step 3: Feature engineering
df['Age_Group'] = pd.cut(df['Age'], bins=[0, 30, 50, 100], labels=['Young', 'Middle', 'Senior'])
df['Performance'] = np.where(df['Score'] >= 80, 'High', 'Low')

# Step 4: GroupBy analysis
print("GroupBy summary:")
print(df.groupby('Age_Group')['Salary'].agg(['mean', 'min', 'max']))

# Step 5: Outlier detection
Q1 = np.percentile(df['Salary'], 25)
Q3 = np.percentile(df['Salary'], 75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df['Salary_Clipped'] = np.clip(df['Salary'], lower, upper)

print("\nOutliers removed using clipping")
print("Final dataset shape:", df.shape)
```

```python
# NP50. Build an end-to-end pipeline using NumPy + Pandas.
np.random.seed(42)
n = 500

# Step 1: Generate raw data
df = pd.DataFrame({
    'Income': np.random.randn(n) * 30000 + 60000,
    'Age': np.random.randint(18, 70, n),
    'Score': np.random.randn(n) * 15 + 80,
    'Category': np.random.choice(['A', 'B', 'C', 'D'], n)
})

print("Initial shape:", df.shape)

# Step 2: Clean invalid values
df['Income'] = np.clip(df['Income'], 20000, 200000)
df['Score'] = np.clip(df['Score'], 0, 100)

# Step 3: Feature engineering
df['Log_Income'] = np.log1p(df['Income'])
df['Income_Squared'] = np.square(df['Income'])
df['Age_Group'] = np.digitize(df['Age'], bins=[0, 25, 45, 65])
df['High_Score'] = np.where(df['Score'] >= 80, 1, 0)

# Step 4: Statistical analysis
print("\nStatistical summary:")
print(df.groupby('Category')['Income'].agg(['mean', 'std', 'count']))

# Step 5: Outlier removal using IQR
arr = df['Income'].values
Q1, Q3 = np.percentile(arr, [25, 75])
IQR = Q3 - Q1
lo, hi = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
df['Income_Clipped'] = np.clip(arr, lo, hi)

# Step 6: ML feature matrix
features = ['Income', 'Age', 'Score', 'Log_Income', 'Income_Squared']
X = df[features].to_numpy(dtype=np.float32)
y = df['High_Score'].to_numpy(dtype=np.float32)

print("\nFeature matrix shape:", X.shape)
print("Label vector shape:", y.shape)

# Step 7: Export
df.to_csv('end_to_end_analysis.csv', index=False)
print("\n Analysis complete! Saved to 'end_to_end_analysis.csv'")
```
