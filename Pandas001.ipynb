```python
# 1. Create a Series using a list of 5 numbers
series1 = pd.Series([10, 20, 30, 40, 50])
print("Series from list:")
print(series1)
print("\n")
```

```python
# 2. Create a Series using a dictionary of student marks
marks_dict = {'Alice': 85, 'Bob': 92, 'Charlie': 78, 'Diana': 88}
series2 = pd.Series(marks_dict)
print("Series from dictionary (student marks):")
print(series2)
print("\n")
```

```python
# 3. Create a DataFrame using a list of dictionaries
data = [
    {'Name': 'John', 'Age': 25, 'City': 'New York'},
    {'Name': 'Emma', 'Age': 30, 'City': 'London'},
    {'Name': 'Mike', 'Age': 35, 'City': 'Paris'}
]
df = pd.DataFrame(data)
print("DataFrame from list of dictionaries:")
print(df)
print("\n")
```

```python
# 4. Add a new column using scalar values
df['Country'] = 'USA'  # Adds scalar value to all rows
print("After adding scalar column 'Country':")
print(df)
print("\n")
```

```python
# 5. Add a new column using a Series
salary_series = pd.Series([50000, 60000, 55000])
df['Salary'] = salary_series
print("After adding Series column 'Salary':")
print(df)
print("\n")
```

```python
# 6. Rename one column and multiple columns
# Rename single column
df.rename(columns={'City': 'Location'}, inplace=True)
print("After renaming 'City' to 'Location':")
print(df)
print("\n")

# Rename multiple columns
df.rename(columns={'Name': 'Full Name', 'Age': 'Years'}, inplace=True)
print("After renaming multiple columns:")
print(df)
print("\n")
```

```python
# 7. Use loc to select rows
# Create a fresh DataFrame for better demonstration
df2 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Score': [85, 92, 78, 88, 95],
    'Grade': ['B', 'A', 'C', 'B', 'A']
})

print("Original DataFrame for loc/iloc examples:")
print(df2)
print("\n")

# Single row
print("Single row (index 2):")
print(df2.loc[2])
print("\n")

# Multiple rows
print("Multiple rows (indices 1, 3):")
print(df2.loc[[1, 3]])
print("\n")

# Conditional rows
print("Rows where Score > 85:")
print(df2.loc[df2['Score'] > 85])
print("\n")
```

```python
# 8. Use iloc to select
# First row
print("First row (iloc[0]):")
print(df2.iloc[0])
print("\n")

# Second column
print("Second column (iloc[:, 1]):")
print(df2.iloc[:, 1])
print("\n")

# First three rows and two columns
print("First three rows and first two columns:")
print(df2.iloc[0:3, 0:2])
print("\n")
```

```python
# 9. Find DataFrame shape and dtypes
print("DataFrame shape (rows, columns):")
print(df2.shape)
print("\n")
print("DataFrame data types:")
print(df2.dtypes)
print("\n")
```

```python
# 10. Handle missing values using fillna()
# Create DataFrame with missing values
df_with_nan = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, np.nan, 6],
    'C': ['x', 'y', np.nan, 'z', 'w']
})

print("Original DataFrame with missing values:")
print(df_with_nan)
print("\n")

# Fill NaN with a scalar value
df_filled_scalar = df_with_nan.fillna(0)
print("After filling NaN with 0:")
print(df_filled_scalar)
print("\n")

# Fill NaN with forward fill (propagate previous value)
df_ffill = df_with_nan.fillna(method='ffill')
print("After forward fill:")
print(df_ffill)
print("\n")

# Fill specific columns with different values
df_custom = df_with_nan.copy()
df_custom['A'] = df_custom['A'].fillna(df_custom['A'].mean())
df_custom['B'] = df_custom['B'].fillna(0)
df_custom['C'] = df_custom['C'].fillna('Unknown')
print("After custom fill (mean for A, 0 for B, 'Unknown' for C):")
print(df_custom)
```
